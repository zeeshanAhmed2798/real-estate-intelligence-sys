```python
import numpy as np
import pandas as pd

from sklearn.model_selection import (
    train_test_split,
    KFold,
    GridSearchCV
)

from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import (
    StandardScaler,
    OrdinalEncoder
)

from sklearn.pipeline import Pipeline
from sklearn.metrics import (
    r2_score,
    mean_absolute_error
)

from xgboost import XGBRegressor
```

```python
# Numerical columns
num_cols = [
    'bedrooms',
    'baths',
    'area_sqft',
    'kitchens',
    'store_rooms',
    'gym',
    'luxury_score',
    'furnishing_score',
    'floors'
]

# Ordinal columns
ordinal_cols = [
    'agePossession'
]

# Binary columns
binary_cols = [
    'property_type',
    'has_servant_room'
]
```

```python
preprocessor = ColumnTransformer(
    transformers=[

        ('num',
         StandardScaler(),
         num_cols),

        ('ord',
         OrdinalEncoder(),
         ordinal_cols),

        ('bin',
         'passthrough',
         binary_cols)

    ],
    remainder='drop'
)
```

```python
X = df[
    num_cols +
    ordinal_cols +
    binary_cols
]

y = df['price']
```

```python
y_transformed = np.log1p(y)
```

```python
X_train, X_test, y_train, y_test = train_test_split(
    X,
    y_transformed,
    test_size=0.2,
    random_state=42
)
```

```python
pipeline = Pipeline([

    ('preprocessor', preprocessor),

    ('regressor',
     XGBRegressor(
         objective='reg:squarederror',
         random_state=42
     ))
])
```

```python
param_grid = {

    'regressor__n_estimators': [200, 300],

    'regressor__max_depth': [4, 6],

    'regressor__learning_rate': [0.05, 0.1],

    'regressor__subsample': [0.8, 1.0],

    'regressor__colsample_bytree': [0.8, 1.0]
}
```

```python
kfold = KFold(
    n_splits=10,
    shuffle=True,
    random_state=42
)
```

```python
search = GridSearchCV(

    pipeline,

    param_grid,

    cv=kfold,

    scoring='r2',

    n_jobs=-1,

    verbose=4
)
```

```python
search.fit(X_train, y_train)
```

```python
search.best_score_
```

```python
search.best_score_
```

```python
final_pipe = search.best_estimator_
```

```python
y_pred_log = final_pipe.predict(X_test)
```

```python
y_pred = np.expm1(y_pred_log)

y_actual = np.expm1(y_test)
```

```python
mean_absolute_error(
    y_actual,
    y_pred
)
```

```python
r2_score(
    y_actual,
    y_pred
)
```
